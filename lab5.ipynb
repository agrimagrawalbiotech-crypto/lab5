Q3. Print the RNA version of hbb, and then separately of the last twelve bases of hbb - that is, the last twelve bases with every T
replaced by U. Then state in one sentence whether you took the slice first or made the substitution first, and whether the
other order would have given the same answer.
Hint: Try it both ways before you answer. The sentence is worth more marks than the output.


```python
# replacing all Ts with Us and  then printing last 12 bases
print(hbb.replace("T","U"), hbb[-12:len(hbb)], sep="\n")
```

```python
# printing last 12 bases first and then replacing T with U
print(hbb[-12:len(hbb)], hbb.replace("T","U"), sep="\n")
```

```python
rna_hbb = hbb.replace("T","U")
print(rna_hbb, rna_hbb[-12:len(rna_hbb)], sep="\n")

rna_hbb1 = hbb
print(rna_hbb1[-12:len(rna_hbb1)], rna_hbb1.replace("T","U"), sep="\n")
```

if we combine the replacing and slicing operations in the print function, it will not store the values of either command in memory and instead only give the output by applying the functions to original hbb individually. this is shown using the hbb variable itself.

However, if we update the value of hbb, or create a new variable rna_hbb and then apply the necessary operations, it will give the true replaced value only when replacing is done first. this is done by using the rna_hbb variable.

In the case of replacement being done AFTER the slicing, the original (DNA) value of hbb will show up instead of the RNA value. This is shown by using the rna_hbb1 variable
